```python
import pandas as pd

# Ensure 'date' is in datetime format
df["date"] = pd.to_datetime(df["date"], errors='coerce')

# Fill missing numerical values with median
df["views"] = df.groupby("grouped_category")["views"].transform(lambda x: x.fillna(x.median()))
df["likes"] = df.groupby("grouped_category")["likes"].transform(lambda x: x.fillna(x.median()))
df["comments"] = df.groupby("grouped_category")["comments"].transform(lambda x: x.fillna(x.median()))

# Fill missing date values using mode or dataset-wide min date
def fill_date(group):
    if group.isna().all():  # If all dates in the group are NaN
        return group.fillna(df["date"].min())  # Fill with the earliest date in dataset
    mode_value = group.mode()
    return group.fillna(mode_value[0] if not mode_value.empty else group.min())

df["date"] = df.groupby("grouped_category")["date"].transform(fill_date)

# Check remaining missing values
print(df.isna().sum())  # Should show 0 missing values for 'date'
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Replace infinite values with NaN
df["engagement_rate"] = df["engagement_rate"].replace([float('inf'), -float('inf')], pd.NA)

# Drop rows with NaN values in 'engagement_rate' (optional)
df = df.dropna(subset=["engagement_rate"])

# Plot histogram
plt.figure(figsize=(8, 5))
sns.histplot(df["engagement_rate"], bins=30, kde=True, color="blue")
plt.title("Engagement Rate Distribution")
plt.xlabel("Engagement Rate")
plt.ylabel("Frequency")
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(df['engagement_rate'].dropna())

print(f"ADF Statistic: {result[0]}")
print(f"p-value: {result[1]}")

if result[1] < 0.05:
    print("The data is stationary (No need for differencing).")
else:
    print("The data is non-stationary (Differencing needed).")
```

```python
from statsmodels.tsa.arima.model import ARIMA

# Fit ARIMA model (p=1, d=1, q=1 as a starting point)
model = ARIMA(df['engagement_rate'], order=(1,1,1))  # Adjust order as needed
model_fit = model.fit()

# Forecast next 30 days
forecast = model_fit.forecast(steps=30)

# Plot results
plt.figure(figsize=(10,5))
plt.plot(df.index, df['engagement_rate'], label="Actual Engagement")
plt.plot(pd.date_range(df.index[-1], periods=30, freq='D'), forecast, label="Predicted Engagement", linestyle="dashed", color="red")
plt.xlabel("Date")
plt.ylabel("Engagement Rate")
plt.legend()
plt.title("ARIMA Engagement Rate Forecast")
plt.show()
```

```python
conn = pymysql.connect(
    host="localhost",
    user="root",
    password="Abhishek"
)
cursor = conn.cursor()
cursor.execute("CREATE DATABASE IF NOT EXISTS Mediacontent;")
cursor.execute("USE Mediacontent;")
print("Database Mediacontent is now selected!")
```

```python
# Drop existing tables (optional)
cursor.execute("DROP TABLE IF EXISTS fact_content, dim_date, dim_content, dim_engagement")

# Create dim_date table
cursor.execute("""
    CREATE TABLE dim_date (
        date_id INT AUTO_INCREMENT PRIMARY KEY,
        publish_date DATE,
        day INT,
        month INT,
        year INT,
        day_of_week VARCHAR(20)
    )
""")

# Create dim_content table
cursor.execute("""
    CREATE TABLE dim_content (
        content_id INT AUTO_INCREMENT PRIMARY KEY,
        headline VARCHAR(255),
        category VARCHAR(100)
    )
""")

# Create dim_engagement table
cursor.execute("""
    CREATE TABLE dim_engagement (
        engagement_id INT AUTO_INCREMENT PRIMARY KEY,
        views INT,
        likes INT,
        comments INT,
        engagement_rate FLOAT
    )
""")

# Create fact_content table
cursor.execute("""
    CREATE TABLE fact_content (
        fact_id INT AUTO_INCREMENT PRIMARY KEY,
        content_id INT,
        date_id INT,
        engagement_id INT,
        FOREIGN KEY (content_id) REFERENCES dim_content(content_id),
        FOREIGN KEY (date_id) REFERENCES dim_date(date_id),
        FOREIGN KEY (engagement_id) REFERENCES dim_engagement(engagement_id)
    )
""")

print("Tables created successfully!")
```

```python
for _, row in df.iterrows():
    # Insert into dim_content
    cursor.execute("INSERT INTO dim_content (headline, grouped_category) VALUES (%s, %s)", 
                   (row["headline"], row["grouped_category"]))
    content_id = cursor.lastrowid

    # Insert into dim_date
    cursor.execute("INSERT INTO dim_date (date, day, month, year, day_of_week) VALUES (%s, %s, %s, %s, %s)", 
                   (row["date"], row["day"], row["month"], row["year"], row["day_of_week"]))
    date_id = cursor.lastrowid

    # Calculate engagement rate
    engagement_rate = (row["likes"] + row["comments"]) / row["views"] if row["views"] > 0 else 0

    # Insert into dim_engagement
    cursor.execute("INSERT INTO dim_engagement (views, likes, comments, engagement_rate) VALUES (%s, %s, %s, %s)", 
                   (row["views"], row["likes"], row["comments"], engagement_rate))
    engagement_id = cursor.lastrowid

    # Insert into fact_content
    cursor.execute("INSERT INTO fact_content (content_id, date_id, engagement_id) VALUES (%s, %s, %s)", 
                   (content_id, date_id, engagement_id))

print("Data inserted successfully!")
```

```python
#Find Content With Highest Engagement Rate
query = """
SELECT dc.headline, dc.grouped_category, de.engagement_rate
FROM fact_content fc
JOIN dim_content dc ON fc.content_id = dc.content_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
ORDER BY de.engagement_rate DESC
LIMIT 5;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
```

```python
#View Engagement Data for a Specific Category
query = """
SELECT dc.grouped_category, SUM(de.views) AS total_views, SUM(de.likes) AS total_likes, SUM(de.comments) AS total_comments
FROM fact_content fc
JOIN dim_content dc ON fc.content_id = dc.content_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
WHERE dc.grouped_category = 'Technology'
GROUP BY dc.grouped_category;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
```

```python
#Find Trending Content Published on Weekends
query = """
SELECT dc.headline, dd.day_of_week, de.views
FROM fact_content fc
JOIN dim_content dc ON fc.content_id = dc.content_id
JOIN dim_date dd ON fc.date_id = dd.date_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
WHERE dd.day_of_week IN ('Saturday', 'Sunday')
ORDER BY de.views DESC
LIMIT 5;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
```

```python
# Find Monthly Trends
query = """
SELECT dd.month, COUNT(fc.fact_id) AS total_posts, SUM(de.views) AS total_views
FROM fact_content fc
JOIN dim_date dd ON fc.date_id = dd.date_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
GROUP BY dd.month
ORDER BY total_views DESC;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
```

```python
# Find Most Commented Video
query = """
SELECT dc.headline, de.comments
FROM fact_content fc
JOIN dim_content dc ON fc.content_id = dc.content_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
ORDER BY de.comments DESC
LIMIT 1;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
```

```python
#Find Categories with Highest Engagement Rate
query = """
SELECT dc.grouped_category, AVG(de.engagement_rate) AS avg_engagement_rate
FROM fact_content fc
JOIN dim_content dc ON fc.content_id = dc.content_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
GROUP BY dc.grouped_category
ORDER BY avg_engagement_rate DESC
LIMIT 5;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
```

```python
#Find Day of the Week with Most Content Engagement
query = """
SELECT dd.day_of_week, SUM(de.views) AS total_views, SUM(de.likes) AS total_likes, SUM(de.comments) AS total_comments
FROM fact_content fc
JOIN dim_date dd ON fc.date_id = dd.date_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
GROUP BY dd.day_of_week
ORDER BY total_views DESC;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
```

```python
#Find Most Popular Content Published in the Last 30 Days
query = """
SELECT dc.headline, dd.date, de.views
FROM fact_content fc
JOIN dim_content dc ON fc.content_id = dc.content_id
JOIN dim_date dd ON fc.date_id = dd.date_id
JOIN dim_engagement de ON fc.engagement_id = de.engagement_id
WHERE dd.date >= DATE_SUB(CURDATE(), INTERVAL 30 DAY)
ORDER BY de.views DESC
LIMIT 5;
"""
cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)

```
